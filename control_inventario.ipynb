## **Order Forecast**


### **Demanda**


```python
ventas_articulos.loc[:, "DEMANDA MENSUAL"] = np.ceil(
    ventas_articulos.CANTIDAD / 30
).astype(int)
ventas_articulos
```

```python
ventas_articulos.loc[:, "DEMANDA DIARIA"] = (
    ventas_articulos.CANTIDAD / days
).round(2)
```

```python
ventas_articulos.sort_values(by="DEMANDA DIARIA", ascending=False)
```

### **Cobertura**


```python
dias_cobertura = 270
ventas_articulos.loc[:, "COBERTURA (días)"] = dias_cobertura
ventas_articulos
```

### **Stock Seguridad**


```python
dias_colchon = 15

ventas_articulos.loc[:, "STOCK SEGURIDAD"] = np.ceil(
    ventas_articulos["DEMANDA DIARIA"] * dias_colchon
).astype(int)

ventas_articulos.sort_values(by="DEMANDA DIARIA", ascending=False)
```

```python
# ventas_articulos.to_excel("output.xlsx")
```

### **Punto de reorden**


```python
lead_time = 30

ventas_articulos["PUNTO DE REORDEN"] = (
    ventas_articulos["DEMANDA DIARIA"] * lead_time +
    ventas_articulos["STOCK SEGURIDAD"]
).round(0)

ventas_articulos.sort_values(by="DEMANDA DIARIA", ascending=False)
```

### **Cantidad a ordenar**


```python
ventas_articulos = ventas_articulos.merge(inventario_disponible[['CODIGO', 'EXISTENCIA']], 
                                          on='CODIGO',
                                          how='left')

# Si un articulo vendido no aparece en el inventario, asumimos existencia = 0
ventas_articulos['EXISTENCIA'] = ventas_articulos['EXISTENCIA'].fillna(0)
```

```python
# Cantidad necesaria para cubrir la demanda del periodo, menos lo que ya hay en existencia.
# Si la existencia ya cubre la demanda, no se debe sugerir una compra negativa: se usa clip(lower=0)
# en lugar de abs(), que invertia incorrectamente los sobrantes en "a ordenar".
ventas_articulos['CANTIDAD A ORDENAR'] = np.ceil(
    ((ventas_articulos['DEMANDA DIARIA'] * dias_cobertura) - ventas_articulos['EXISTENCIA']).clip(lower=0)
).astype(int)
```

```python
ventas_articulos.sort_values(by="DEMANDA DIARIA", ascending=False)
```

```python
forecast = ventas_articulos.sort_values(by="DESCRIPCION", ascending=True)
forecast.to_html(f'{marca}_output.html')
```
